### 2. Análisis descriptivo global

Se realiza un análisis estadístico de las variables mediante `describe()`, con el objetivo de identificar sus principales características, como la media, desviación estándar, valores mínimos, máximos y cuartiles.

```python
df.describe()
```

```python

# Calcular los cuartiles y el rango intercuartílico (IQR)
Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)
IQR = Q3 - Q1

# Definir los límites para identificar outliers
Limite_inferior = Q1 - 1.5 * IQR
Limite_superior = Q3 + 1.5 * IQR

# Filtrar los valores atípicos
outliers = df[
    (df["Amount"] < Limite_inferior) |
    (df["Amount"] > Limite_superior)
]

# Mostrar resultados
print("Límite inferior:", Limite_inferior)
print("Límite superior:", Limite_superior)
print("Número de outliers:", len(outliers))
print("Porcentaje de outliers:", round(len(outliers) / len(df) * 100, 2), "%")
```

### Interpretación del análisis descriptivo

El análisis estadístico confirma que las 30 variables contienen 10.492 observaciones.

La variable `Amount` presenta una media de 91,02 y una mediana de 20,99, lo que indica una distribución asimétrica hacia valores elevados. El importe máximo de 7.712,43 sugiere la existencia de posibles valores atípicos.

Por su parte, la variable `Class` presenta una media de 0,0469, equivalente aproximadamente a un 4,69 % de transacciones fraudulentas, lo que evidencia un desbalance entre ambas clases.

Continuación de la Práctica 1. Se utilizan las características de las transacciones para entrenar los algoritmos; la columna `Class` se conserva exclusivamente para evaluar sus resultados.

## 5. Preparación de las variables sin `Class`

```python

# Crear una copia del DataFrame sin la variable objetivo Class
df_sin_target = df.drop(columns=["Class"]).copy()

# Comprobar las dimensiones del nuevo DataFrame
print("Dimensiones originales:", df.shape)
print("Dimensiones sin target:", df_sin_target.shape)

# Visualizar las primeras filas
df_sin_target.head()
```

## 6. Matriz de características y estandarización

```python
matriz_df_sin_target = df_sin_target.values
```

```python

# Importar la herramienta de estandarización
from sklearn.preprocessing import StandardScaler

# Crear el objeto StandardScaler
scaler = StandardScaler()

# Calcular media y desviación de cada variable
# y transformar los datos
matriz_estandarizada = scaler.fit_transform(
    matriz_df_sin_target
)

# Comprobar las medias
print("Medias:")
print(matriz_estandarizada.mean(axis=0).round(2))

# Comprobar las desviaciones estándar
print("\nDesviaciones estándar:")
print(matriz_estandarizada.std(axis=0).round(2))
```

## 7. DBSCAN — Análisis exploratorio de distancias
El gráfico de distancia al décimo vecino sirve para orientar el valor inicial de `eps`; es un análisis complementario a los requisitos del ejercicio.

```python

from sklearn.neighbors import NearestNeighbors
import numpy as np
import matplotlib.pyplot as plt

# Número de vecinos que vamos a estudiar
n_vecinos = 10

# Calcular los vecinos más cercanos
vecinos = NearestNeighbors(n_neighbors=n_vecinos)
vecinos.fit(matriz_estandarizada)

# Obtener las distancias a los vecinos
distancias, indices = vecinos.kneighbors(
    matriz_estandarizada
)

# Ordenar la distancia al décimo vecino
distancias_ordenadas = np.sort(distancias[:, -1])

# Visualizar las distancias
plt.figure(figsize=(10, 5))
plt.plot(distancias_ordenadas)
plt.title("Distancia al décimo vecino más cercano")
plt.xlabel("Transacciones ordenadas")
plt.ylabel("Distancia euclidiana")
plt.grid(True)
plt.show()
```

## 8. DBSCAN — Modelado y detección de anomalías

```python

# Importar DBSCAN
from sklearn.cluster import DBSCAN
import numpy as np

# Definir el modelo y sus parámetros
modelo_dbscan_1 = DBSCAN(
    eps=5.0,
    min_samples=10,
    metric="euclidean"
)

# Aplicar DBSCAN a los datos estandarizados
labels_1 = modelo_dbscan_1.fit_predict(
    matriz_estandarizada
)

# Contar el número de clústeres encontrados
n_clusters = len(set(labels_1)) - (1 if -1 in labels_1 else 0)

# Contar las observaciones clasificadas como anomalías
n_anomalias = np.sum(labels_1 == -1)

# Mostrar los resultados
print("Número de clústeres:", n_clusters)
print("Número de anomalías:", n_anomalias)
print("Porcentaje de anomalías:",
      round(n_anomalias / len(labels_1) * 100, 2), "%")
```

### 8.1. Evaluación con las etiquetas reales

```python

# Crear una copia del DataFrame original
df_dbscan = df.copy()

# Añadir las etiquetas de los clusters
df_dbscan["cluster_dbscan_1"] = labels_1

# Convertir las etiquetas en predicciones binarias
df_dbscan["pred_dbscan_1"] = np.where(
    df_dbscan["cluster_dbscan_1"] == -1, 1, 0
)

# Comprobar los resultados
df_dbscan[["Class", "cluster_dbscan_1", "pred_dbscan_1"]].head(10)
```

```python

# Importar las herramientas de evaluación
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Definir los valores reales y las predicciones
y_true = df_dbscan["Class"]
y_pred = df_dbscan["pred_dbscan_1"]

# Calcular la matriz de confusión
cm = confusion_matrix(y_true, y_pred, labels=[0, 1])

# Visualizar la matriz
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Legítima", "Fraude"]
)

disp.plot(cmap="Blues")
plt.title("Matriz de confusión - DBSCAN")
plt.grid(False)
plt.show()
```

```python

from sklearn.metrics import precision_score, recall_score, f1_score

# Calcular las métricas de evaluación
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

# Mostrar los resultados en porcentaje
print(f"Precision: {precision:.2%}")
print(f"Recall: {recall:.2%}")
print(f"F1-score: {f1:.2%}")
```

### Conclusiones del modelo DBSCAN

Se implementó el algoritmo DBSCAN con `eps=5`, `min_samples=10` y distancia euclidiana, identificando **3 clústeres y 334 transacciones anómalas (3,18 % del total)**.

La matriz de confusión mostró 159 verdaderos positivos (TP), 9.825 verdaderos negativos (TN), 175 falsos positivos (FP) y 333 falsos negativos (FN).

Las métricas obtenidas fueron:

- **Precision: 47,60 %**, indicando que aproximadamente la mitad de las alertas correspondían a fraudes reales.
- **Recall: 32,32 %**, reflejando que el modelo únicamente identificó alrededor de un tercio de los fraudes existentes.
- **F1-score: 38,50 %**, evidenciando un rendimiento limitado al combinar ambas métricas.

**En conclusión**, aunque DBSCAN logró identificar algunas transacciones fraudulentas, presentó una elevada cantidad de falsos negativos. Por tanto, su rendimiento con los parámetros seleccionados resulta limitado para la detección de fraude financiero. Estos resultados servirán como referencia para compararlo posteriormente con Isolation Forest.


## 9. Isolation Forest — Entrenamiento y puntuación de anomalías

```python

# Importar el algoritmo Isolation Forest
from sklearn.ensemble import IsolationForest

# Definir el modelo
iso_1 = IsolationForest(
    n_estimators=100,
    max_samples="auto",
    random_state=42
)
```

```python

# Entrenar Isolation Forest con los datos estandarizados
iso_1.fit(matriz_estandarizada)

# Obtener las puntuaciones de anomalía
score_anomalia_1 = -iso_1.decision_function(
    matriz_estandarizada
)

# Visualizar las primeras puntuaciones
print(score_anomalia_1[:10])
```

### 9.1. Evaluación del modelo

```python

# Crear una copia del DataFrame original
df_iso = df.copy()

# Agregar las puntuaciones de anomalía
df_iso["score_iso_1"] = score_anomalia_1

# Visualizar los primeros resultados
df_iso[["Class", "score_iso_1"]].head(10)
```

### 9.2. Umbral inicial: percentil 95

```python

# Definir el percentil de anomalía
N = 95

# Calcular el umbral
umbral_1 = np.percentile(score_anomalia_1, N)

# Clasificar las transacciones según el umbral
df_iso["flag_anomalia_1"] = (
    df_iso["score_iso_1"] >= umbral_1
).astype(int)

# Mostrar los resultados
print("Umbral de anomalía:", umbral_1)

print("\nDistribución de predicciones:")
print(df_iso["flag_anomalia_1"].value_counts())
```

Se seleccionó el percentil 95 como umbral inicial de detección de anomalías, clasificando como sospechosas aproximadamente el 5 % de las transacciones con mayores puntuaciones. Este criterio permite priorizar las operaciones más atípicas y establecer un volumen acotado de alertas para su revisión, sin utilizar las etiquetas reales de fraude durante la definición del umbral.

```python

# Definir las etiquetas reales
y_true_iso = df_iso["Class"]

# Definir las predicciones de Isolation Forest
y_pred_iso = df_iso["flag_anomalia_1"]

# Calcular la matriz de confusión
cm_iso = confusion_matrix(
    y_true_iso,
    y_pred_iso,
    labels=[0, 1]
)

# Visualizar la matriz de confusión
disp_iso = ConfusionMatrixDisplay(
    confusion_matrix=cm_iso,
    display_labels=["Legítima", "Fraude"]
)

disp_iso.plot(cmap="Blues")
plt.title("Matriz de confusión - Isolation Forest")
plt.grid(False)
plt.show()
```

```python

# Calcular las métricas de Isolation Forest
precision_iso = precision_score(y_true_iso, y_pred_iso)
recall_iso = recall_score(y_true_iso, y_pred_iso)
f1_iso = f1_score(y_true_iso, y_pred_iso)

# Mostrar resultados
print(f"Precision: {precision_iso:.2%}")
print(f"Recall: {recall_iso:.2%}")
print(f"F1-score: {f1_iso:.2%}")
```

### Conclusiones de Isolation Forest

Se implementó el algoritmo Isolation Forest utilizando 100 árboles y un umbral de anomalía basado en el percentil 95, identificando 525 transacciones sospechosas (5 % del total).

La matriz de confusión mostró 329 verdaderos positivos (TP), 9.804 verdaderos negativos (TN), 196 falsos positivos (FP) y 163 falsos negativos (FN).

El modelo obtuvo una **precision del 62,67 %, un recall del 66,87 % y un F1-score del 64,70 %**.

Estos resultados muestran una capacidad de detección de fraude superior a la obtenida con DBSCAN. Sin embargo, Isolation Forest todavía deja sin identificar aproximadamente un tercio de las operaciones fraudulentas, por lo que presenta limitaciones para su utilización en escenarios financieros reales.

### 9.3. Análisis complementario de sensibilidad: percentil 85

```python

# Definir el nuevo percentil
N = 85

# Calcular el nuevo umbral
umbral_85 = np.percentile(score_anomalia_1, N)

# Generar predicciones con P85
df_iso["flag_anomalia_85"] = (
    df_iso["score_iso_1"] >= umbral_85
).astype(int)

# Etiquetas reales y nuevas predicciones
y_true_iso = df_iso["Class"]
y_pred_85 = df_iso["flag_anomalia_85"]

# Calcular las métricas
precision_85 = precision_score(y_true_iso, y_pred_85)
recall_85 = recall_score(y_true_iso, y_pred_85)
f1_85 = f1_score(y_true_iso, y_pred_85)

# Mostrar resultados
print("Anomalías detectadas:", y_pred_85.sum())
print(f"Precision: {precision_85:.2%}")
print(f"Recall: {recall_85:.2%}")
print(f"F1-score: {f1_85:.2%}")
```

### Evaluación adicional de Isolation Forest — Percentil 85

Se realizó una segunda evaluación utilizando el percentil 85 como umbral de anomalía, clasificando 1.574 transacciones como sospechosas (15 % del conjunto de datos).

El modelo obtuvo una **precision del 27,32 %, un recall del 87,40 % y un F1-score del 41,63 %**.

En comparación con P95, se detectaron 101 fraudes adicionales, reduciendo los falsos negativos de 163 a 62. Sin embargo, los falsos positivos aumentaron de 196 a 1.144.

**En conclusión**, P85 mejora considerablemente la capacidad de detección de fraude, pero reduce la precision y genera muchas más falsas alarmas. Por tanto, P95 ofrece un mejor equilibrio entre ambas métricas, mientras que P85 resulta más adecuado cuando se prioriza la detección del mayor número posible de fraudes.
